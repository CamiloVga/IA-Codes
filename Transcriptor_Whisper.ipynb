```python
!pip install git+https://github.com/openai/whisper.git
!pip install gradio

import whisper
import gradio as gr

model = whisper.load_model("large")

```

```python
def transcribir_audio(audio):
  resultado = model.transcribe(audio)
  return resultado["text"]

interfaz = gr.Interface(
    fn=transcribir_audio,
    inputs=gr.Audio(type="filepath"),  # Cambiado a "filepath"
    outputs="text",
    title="Transcripción de audio con Whisper Large"
)

interfaz.launch()
```
